```python
from datetime import date, timedelta
from pathlib import Path

import pandas as pd

yesterday = date.today() - timedelta(days=1)

# 5 lignes par source pour la date d'hier (data/bronze/source=<id>/date=<hier>/articles.parquet)
files = sorted(Path("./data/bronze").glob(f"source=*/date={yesterday}/articles.parquet"))
df = pd.concat((pd.read_parquet(f).head(5) for f in files), ignore_index=True)
print(f"{yesterday} : {len(files)} sources")
df
```
